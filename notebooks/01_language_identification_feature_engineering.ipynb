A "amostras" de texto precisa ser "transformada" em <b>padrões</b>

<p>Um padrão é um conjunto de características, geralmente representado por um vetor e um conjunto de padrões no formato de tabela. Onde cada linha é um padrão e as colunas as características e, geralmente, na última coluna a <b>classe</b></p>

```python
import random

pre_padroes = []
for frase in ingles:
  pre_padroes.append( [frase, 'inglês'])

for frase in espanhol:
  pre_padroes.append( [frase, 'espanhol'])

for frase in portugues:
  pre_padroes.append( [frase, 'português'])

random.shuffle(pre_padroes)
print(pre_padroes)
```

<h1>Construção dos atributos</h1>
<p>Esse é o coração desse trabalho e que deverá ser desenvolvido por vocês. Pensem em como podemos "medir" cadas frase/sentença e extrair características que melhorem o resultado do processo de identificação.<p>
<p>Após a criação de cada novo atributo, execute as etapas seguintes e registre as métricas da matriz de confusão. Principalmente acurácia e a precisão.</p>

```python
# a entrada é o vetor pre_padroes e a saída desse passo deverá ser "padrões"
import re

def tamanhoMedioFrases(texto):
    palavras = re.split(r"\s", texto)
    tamanhos = [len(s) for s in palavras if len(s) > 0]
    if not tamanhos: return 0
    return sum(tamanhos) / len(tamanhos)

# Contagem de palavras base
def contagem_palavras_pt(texto):
    comuns = {'o', 'a', 'os', 'as', 'de', 'do', 'da', 'em', 'no', 'na', 'para', 'com', 'que', 'é', 'um', 'uma'}
    return sum(1 for p in texto.lower().split() if p in comuns)

def contagem_palavras_es(texto):
    comuns = {'el', 'la', 'los', 'las', 'de', 'del', 'en', 'por', 'con', 'que', 'es', 'y', 'un', 'una'}
    return sum(1 for p in texto.lower().split() if p in comuns)

def contagem_palavras_en(texto):
    comuns = {'the', 'a', 'an', 'of', 'in', 'to', 'for', 'with', 'that', 'is', 'and', 'my', 'i'}
    return sum(1 for p in texto.lower().split() if p in comuns)

# Símbolos Especiais
def contagem_simbolos_pt(texto):
    return len(re.findall(r'[çãõâêô]', texto.lower()))

def contagem_simbolos_es(texto):
    return len(re.findall(r'[ñ¿¡]', texto.lower()))

# Dois Símbolos Sucessivos
def contagem_bigramas_en(texto):
    return len(re.findall(r'(th|wh|oo|ee)', texto.lower()))

def contagem_bigramas_es(texto):
    return len(re.findall(r'(ll)', texto.lower()))

# Sufixos
def contagem_sufixos(texto):
    texto_lower = texto.lower()
    pt = len(re.findall(r'(ção|ções|dade)\b', texto_lower))
    es = len(re.findall(r'(ción|ciones|dad)\b', texto_lower))
    en = len(re.findall(r'(tion|ing|ly)\b', texto_lower))
    return pt, es, en

def extraiCaracteristicas(frase):
    texto_original = frase[0]

    # Limpeza de pontuação apenas para a contagem de palavras exatas
    texto_limpo = re.sub(re.compile(r'[^\w\s]', re.UNICODE), ' ', texto_original)

    # Extração das features
    c_tamanho = tamanhoMedioFrases(texto_limpo)
    c_pt = contagem_palavras_pt(texto_limpo)
    c_es = contagem_palavras_es(texto_limpo)
    c_en = contagem_palavras_en(texto_limpo)

    s_pt = contagem_simbolos_pt(texto_original)
    s_es = contagem_simbolos_es(texto_original)

    b_en = contagem_bigramas_en(texto_original)
    b_es = contagem_bigramas_es(texto_original)

    suf_pt, suf_es, suf_en = contagem_sufixos(texto_original)

    # Montagem do vetor de padrões (11 características + classe)
    padrao = [
        c_tamanho, c_pt, c_es, c_en,
        s_pt, s_es,
        b_en, b_es,
        suf_pt, suf_es, suf_en,
        frase[1]
    ]
    return padrao


def geraPadroes(frases):
    return [extraiCaracteristicas(frase) for frase in frases]

padroes = geraPadroes(pre_padroes)

colunas = [
    'Tamanho_Medio', 'Palavras_PT', 'Palavras_ES', 'Palavras_EN',
    'Simbolos_PT', 'Simbolos_ES',
    'Bigramas_EN', 'Bigramas_ES',
    'Sufixos_PT', 'Sufixos_ES', 'Sufixos_EN',
    'Idioma'
]

dados = pd.DataFrame(padroes, columns=colunas)
display(dados)
```

<h2>Treinando o modelo com SVM</h1>
<p>Separando o conjunto de treinamento do conjunto de testes</p>

```python
from sklearn.model_selection import train_test_split
import numpy as np


#from sklearn.metrics import confusion_matrix

vet = np.array(padroes)
classes = vet[:,-1]         # classes = [p[-1] for p in padroes]
#print(classes)
padroes_sem_classe = vet[:,0:-1]
#print(padroes_sem_classe)
X_train, X_test, y_train, y_test = train_test_split(padroes_sem_classe, classes, test_size=0.25, stratify=classes)
```

Com os conjuntos separados, podemos "treinar" o modelo usando a SVM.

```python
from sklearn import svm
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report

treinador = svm.SVC()  #algoritmo escolhido
modelo = treinador.fit(X_train, y_train)

#
# score com os dados de treinamento
acuracia = modelo.score(X_train, y_train)
print("Acurácia nos dados de treinamento: {:.2f}%".format(acuracia * 100))

#
# melhor avaliar com a matriz de confusão
y_pred = modelo.predict(X_train)
cm = confusion_matrix(y_train, y_pred)
print(cm)
print(classification_report(y_train, y_pred))

#
# com dados de teste que não foram usados no treinamento
print('métricas mais confiáveis')
y_pred2 = modelo.predict(X_test)
cm = confusion_matrix(y_test, y_pred2)
print(cm)
print(classification_report(y_test, y_pred2))
```
